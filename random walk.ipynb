# 2 dimension random walk generator

### One Dimension Version

Path is from left to right only

```python
import numpy as np
import matplotlib.pyplot as plt

position = 0
steps = 100000
walk = [position]
for i in range(steps): 
    step = 1 if np.random.randint(0, 2) else -1
    position += step
    walk.append(position)

plt.plot(walk);
```

## 2 Dimensional Version

Path should be able to go both left and right at any point, starting at the origin

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.collections import LineCollection 

xpos, ypos = 0, 0
nsteps = 100000
walk = [(xpos, ypos)]
for i in range(nsteps):
    xstep = 1 if np.random.randint(0, 2) else -1
    ystep = 1 if np.random.randint(0, 2) else -1
    xpos, ypos = walk[-1]
    xpos += xstep
    ypos += ystep
    walk.append((xpos, ypos))

x, y = zip(*walk)


fig, ax = plt.subplots(figsize=(10,10))
fig.patch.set_facecolor('#ffffff')

ax.grid(True)
ax.set_aspect('equal')
ax.set(xlim=(min(x), max(x)), ylim=(min(y), max(y)))
ax.spines['bottom'].set_position('zero')
ax.spines['left'].set_position('zero')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.plot(x, y)
```
