```python
import matplotlib.pyplot as plt
import sympy as sp
import pandas as pd
```

```python
d = list()
v_initial = 12.9
v = [x * 0.1 for x in range(-130, 130)]
for i in v: 
    displacement = sp.symbols('displacement')
    expr = sp.Eq(i**2, v_initial**2 - 2 * 9.80 * displacement)
    sol = sp.solve(expr, displacement)
    value = float(sol[0])
    d.append(value)

print(d)
```

```python
time = [x * 0.1 for x in range(0, 30)]
# print(time)
dasplacement = sp.symbols('dasplacement')
das = list()
for t in time:
    expr2 = sp.Eq(dasplacement, v_initial * t - 0.5 * 9.80 * t**2)
    sol2 = sp.solve(expr2, dasplacement)
    value2 = float(sol2[0])
    das.append(value2)

print(das)


    
```

```python
plt.plot(v, d)
plt.xlabel('velocity (m/s)')
plt.ylabel('displacement (m)')
plt.title('Velocity vs Displacement')
plt.show()
```

```python
plt.plot(time, das)
plt.xlabel('time (s)')
plt.ylabel('displacement (m)')
plt.title('Dophin Time vs Displacement')
```
